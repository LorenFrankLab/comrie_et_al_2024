```python
# stats
pvals = {}      

subject_ids = ['senor', 'chimi', 'j16', 'wilbur', 'peanut', 'allrats']

segs = ['is_first_seg_of_trial','is_last_seg_of_trial']
is_switch = [True,False]

for subject_id in subject_ids:
    print(subject_id.upper())
    pvals[subject_id] = {}
    for switch in is_switch:
        pvals[subject_id][switch] = {}
        for seg in segs:
            # read in local nonlocal data
            file_path = f'{DATA_DIR}/figs26/'
            file_name = f'theta_{subject_id}_switch{switch}_seg{seg}True_electrode_cc_values_phase_data_'
            
            phase_local = np.load(file_path+file_name+'local.npy')
            phase_nonlocal = np.load(file_path+file_name+'nonlocal.npy')
            print(f'\n\nloaded local and nonlocal phase data for {subject_id}, switch {switch}, seg {seg} true, electrode cc theta phase values in radians\n')
            
            # kuiper's test, nonparametric to see if distributions differ from ea other, this repo dep'd
            try:
                test_statistic, p_value = circ.tests.kuiper(phase_local, phase_nonlocal)
            except Exception as e:
                print('exception in kuiper test: ', e)
                test_statistic, p_value = np.nan, np.nan
            print(f"\nKuiper's test statistic: {'{:.9e}'.format(test_statistic)}")
            print(f"P-value: {'{:.9e}'.format(p_value)}")
            # another version
            try:
                result = circular.stats.kuiper_two(phase_local, phase_nonlocal, return_all=True)
            except Exception as e:
                print('exception in kuiper_two test: ', e)
                result = [np.nan]
            print(f"\nP-value: {'{:.9e}'.format(result[0])}")
            
            print(f"Sample size local: {len(phase_local)}")
            print(f"Sample size nonlocal: {len(phase_nonlocal)}")
            
            pvals[subject_id][switch][seg] = [result[0]]
```
